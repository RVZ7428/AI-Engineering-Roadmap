```python
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 88}
]
# 온도가 90 이상인 기계의 ID만 추출하기
```

```python
#1 for문
result=[]
for machine in machines:
    if machine["temperature"]>=90:
        result.append(machine["id"])
print(result)
```

```python
#2 리스트 컴프리헨현
result=[
    machine["id"]
    for machine in machines if machine["temperature"]>=90
]
print(result)
```

```python
#3 filter, map
dangerous_mahines=filter(
    lambda machine : machine["temperature"]>=90,
    machines
)

result=list(
    map(lambda machine : machine["id"], dangerous_mahines)
)
print(result)
```

```python
#문제1 for문으로 추출 온도 80이상 기계ID
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 80},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 88}
]
result=[]
for machine in machines:
    if machine["temperature"]>=80:
        result.append(machine["id"])
print(result)
```

```python
#문제2 리스트 컴프리헨션 점수 70 이상 추출
scores = [55, 70, 82, 64, 91, 70, 48]
result=[score for score in scores if score>=70]
print(result)
```

```python
#문제3 filter()와 map() 조합 숫자 20 이상인 숫자 선택후 3 곱하기
numbers = [5, 10, 20, 25, 30]
num20=filter(
    lambda number : number>=20, numbers
)
num3=list(
    map(lambda number : number*3, num20)
)
print(num3)
```

```python
#문제4 함수 만들기
#함수 이름: get_hot_machine_ids(machines)
# 온도 90이상인 기계 선택, 선택한ID리스트 반환 return 사용
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 80},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 88}
]
def get_hot_machine_ids(machines):
    result=[]
    for machine in machines:
        if machine["temperature"]>=90:
            result.append(machine["id"])
    return result
```
