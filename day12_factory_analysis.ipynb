```python
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 85},
    {"id": 103, "temperature": 100},
    {"id": 104, "temperature": 110},
    {"id": 105, "temperature": 95}
]
```

온도	상태
80 미만	NORMAL
80 이상, 100 미만	CHECK
100 이상	DANGER
① get_machine_status(temperature)
온도를 입력받아 NORMAL, CHECK, DANGER 중 하나를 반환

② analyze_machines(machines)
각 기계의 상태를 판정하고 분석 결과를 구성

③ get_dangerous_machine_ids(machines)
위험 기계의 ID만 추출

④ calculate_average_temperature(machines)
평균 온도 계산

⑤ main()
각 함수를 호출하고 분석 결과를 최종 출력

```python
#실습1 상태 팑어 함수 작성
def get_machine_status(temperature):
        if temperature<80:
            return "NORMAL"
        elif temperature<100:
            return "CHECK"
        else:
            return "DANGER"
print(get_machine_status(72))   # NORMAL
print(get_machine_status(85))   # CHECK
print(get_machine_status(100))  # DANGER
```

# 실습2 전체 기계 분석하기
요구사항
1. 각 기계의 온도를 get_machien_status()에 전달
2. 분석 결과에 기계 ID, 온도, 상태를 저장
3. 완성도니 분석 결과 리스트를 return한다


```python
def analyze_machines(machines):
    for machine in machines:
        machine["status"] = get_machine_status(machine["temperature"])
    return machines
    
machines=analyze_machines(machines)
print(machines)
```

실습 3. 위험 기계 ID 추출하기
실습 2에서 상태가 추가된 machines를 사용해.

조건
- 함수 이름: get_dangerous_machine_ids(machines)
- 상태가 "DANGER"인 기계만 찾는다.
- 해당 기계의 "id"만 리스트에 담는다.
- 최종적으로 ID 리스트를 반환한다.

```python
def get_dangerous_machine_ids(machines):
    ids=[]
    for machine in machines:
        if(machine["status"]=="DANGER"):
            ids.append(machine["id"])
    return ids
print(get_dangerous_machine_ids(machines))
```

실습 4. 평균 온도 계산하기
조건
- 함수 이름: calculate_average_temperature(machines)
- 모든 기계의 온도를 더한 뒤 기계 수로 나눈다.
- 기계가 한 대도 없다면 0을 반환한다.
- 평균 온도를 반환한다.

```python
def calculate_average_temperature(machines):
    machine_temp_sum=0
    for machine in machines:
        machine_temp_sum+=machine["temperature"]
    machine_mean=machine_temp_sum/len(machines)
    return machine_mean
print(calculate_average_temperature(machines))
```

```python
# 전체 결과 출력하
def main():
    analyzed_machines = analyze_machines(machines)

    dangerous_ids = get_dangerous_machine_ids(analyzed_machines)
    average_temperature = calculate_average_temperature(analyzed_machines)

    temperatures = []
    for machine in analyzed_machines:
        temperatures.append(machine["temperature"])

    max_temperature = max(temperatures) if temperatures else 0

    print("===== 공장 기계 상태 분석 =====")

    for machine in analyzed_machines:
        print(
            f"기계 ID: {machine['id']} | "
            f"온도: {machine['temperature']} | "
            f"상태: {machine['status']}"
        )

    print("위험 기계 ID:", dangerous_ids)
    print("평균 온도:", average_temperature)
    print("최고 온도:", max_temperature)
    print("전체 기계 수:", len(analyzed_machines))
    print("위험 기계 수:", len(dangerous_ids))


main()
```
