```python
# 데이터 변환
numbers=[1, 2, 3, 4, 5]
result=[]

for number in numbers:
    result.append(number * 2)

print(result)
```

```python
result=[number*2 for number in numbers]
# [결과로 만들 값 for 하나씩 꺼낼 변수 in 원본 데이터]
print(result)
```

```python
#2 변환 + 조건
numbers=[1, 2, 3, 4, 5]
result=[number*2 for number in numbers if number >=3]
print(result)
```

```python
#문제1 리스트 변환 list Comprehension 사용 *2
numbers=[10, 20, 30, 40, 50]
result=[number*2 for number in numbers]
print(result)
```

```python
#문제2 조건+변환
# 다음 점수에서 80점 이상인 점수만 골라서 5점을 추가해라
scores = [65, 82, 91, 73, 88, 55, 100]
result=[score+5 for score in scores if score>=80]
print(result)
```

```python
#문제3 Dictionaty 데이터 변환 각 학생 점수만 뽑아서 새로운 리스트를 만들어라
students = [
    {"name": "철수", "score": 95},
    {"name": "영희", "score": 82},
    {"name": "민수", "score": 73},
    {"name": "지수", "score": 58}
]
result=[student["score"] for student in students] 
print(result)
```

```python
#문제4 Dictionary 학생 이름만 뽑기
students = [
    {"name": "철수", "score": 95},
    {"name": "영희", "score": 82},
    {"name": "민수", "score": 73},
    {"name": "지수", "score": 58}
]
result=[student["name"] for student in students]
print(result)
```

```python
#문제5 데이터 가공 온도 80이상인 기계의 온도만 뽑아서 5를 더한 새로운 리스트
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
result=[machine["temperature"]+5 for machine in machines if machine["temperature"]>= 80]
print(result)
```

```python
#문제6 함수 온도가 80 이상인 기계만 골라서, 온도에 5를 더한 값을 리스트로 반환한다.
def get_high_temperature_scores(machines):
    # 작성
    result=[machine["temperature"]+5 for machine in machines if machine["temperature"]>= 80]
    return result

machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]

result = get_high_temperature_scores(machines)

print(result)
```
