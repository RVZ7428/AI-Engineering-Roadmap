```python
#1 map() 모든 데이터를 변환하기
numbers=[1, 2, 3, 4, 5]
result=list(map(lambda x:x +10, numbers))
print(result)
```

map() : 데이터 각각에 함수를 적용해.

lambda x: x + 10 : 숫자 하나를 받아 10을 더해.

list() : 변환된 결과를 리스트로 만들어.

```python
result=[x+10 for x in numbers]
print(result)
```

```python
#2 filter() 조건에 맞는 데이터만 남기기
numbers = [1, 2, 3, 4, 5, 6]
result=list(filter(lambda x: x%2==0, numbers))
print(result)
```

```python
#3 Dictionary 데이터에 적용
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105}
]
result = list(
    filter(
        lambda machine: machine["temperature"] >= 100,
        machines
    )
)

print(result)
```

```python
#문제1 map으로 변환하기 *2
numbers = [10, 20, 30, 40, 50]
result=list(map(lambda number: number*2, numbers))
print(result)
```

```python
#문제2 filter()로 추출하기 80이상 추출
scores = [65, 82, 91, 73, 88, 55, 100]
result=list(filter(lambda score:score>80, scores))
print(result)
```

```python
#문제3 map()으로 Dictionary 변환하기 각 기계의 온도 5를 더한 온도 리스트
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105}
]
result=list(
    map(
        lambda machine: machine["temperature"]+5,
        machines))
print(result)
```

```python
#문제4 filter()로 위험 기계 추출하기 온도 100 이상인 기계 추출하기

result=list(
    filter(
        lambda machine: machine["temperature"]>=100,
        machines))
print(result)
```

```python
#문제5 map()과 filter() 조합하기 20이상인 숫자만 선택하고 선택된 숫자에 2를 곱해라
numbers = [10, 15, 20, 25, 30, 35]
lists=list(
    filter(lambda number: number>=20, numbers)
)
print(lists)
result=list(
    map(lambda list:list*2, lists)
)
print(result)
```

```python
#문제6 함수 만들기 온도가 100 이상인 기계들의 ID만 리스트로 반환
def get_dangerous_machine_ids(machines):
    dangerous_machines = filter(
        lambda machine: machine["temperature"] >= 100,
        machines
    )
    result = list(
        map(lambda machine: machine["id"], dangerous_machines)
    )
    return result
```
