```python
#기본 정렬
numbers = [30, 10, 50, 20, 40]
result=sorted(numbers)
print(numbers)
print(result)
```

```python
#내림차순
result=sorted(numbers, reverse=True)
print(result)
```

```python
# Dictionary key 사용
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83}
]
result=sorted(
    machines,
    key=lambda machine: machine["temperature"]
    #각 machine에서 temperature를 꺼내서 그 값을 기준으로 정렬해라
)
print(result)
```

```python
#문제1 숫자 오름차순
numbers = [45, 12, 78, 34, 23, 91]
result=sorted(numbers)
print(result)
```

```python
#문제3 학생 점수 낮은 학생 -> 높은 학생
students = [
    {"name": "철수", "score": 85},
    {"name": "영희", "score": 92},
    {"name": "민수", "score": 78},
    {"name": "지수", "score": 96}
]
result=sorted(
    students,
    key=lambda student : student["score"]
)
print(result)
```

```python
#문제4 학생 점수 내림차순
result=sorted(
    students,
    key=lambda student : student["score"],
    reverse=True
)
print(result)
```

```python
#문제5 기계 온도가 높은 기계부터 정렬
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
result=sorted(
    machines,
    key=lambda machine : machine["temperature"],
    reverse=True
)
print(result)
```

```python
#문제6 함수로 만들기, 함소가 호출되면 온도가 높은 순서대로 정렬된 기계 리스트 변환
def sort_machines_by_temperature(machines):
    result=sorted(
        machines,
        key=lambda machine : machine["temperature"],
        reverse=True
        )
    return result
```
