```python
#1 리스트 컴프리헨션
scores = [95, 82, 71, 60, 88, 43]
new_list=[]
for score in scores:
    if score>=80:
        new_list.append(score)
print(new_list)
```

```python
# [표현식 for 변수 in 반복대상 if 조건]
new_list=[score for score in scores if score>=80]
print(new_list)
```

```python
#2 enumerate()
# 리스트, 튜플, 문자열 같은 순서가 있는 자료형을 입력받아 인덱스와 값을 동시에 반환해 주는 내장 함수
machines = ["A", "B", "C"]
for i, machine in enumerate(machines):
    print(i, machine)
```

```python
# 시작 번호 바꾸기
machines = ["A", "B", "C"]
for i, machine in enumerate(machines, start=1):
    print(i, machine)
```

```python
#3 zip()
names=["철수", "영희", "민수"]
scores=[90, 85, 78]
# 두 리스트를 같이 처리
for name, score in zip(names, scores):
    print(name, score)
```

```python
#4 함수 설계
def calculate_score(score):
    return score + 10
result=calculate_score(80)
print(result)
```

```python
#문제1 리스트 컴프리헨션
# 다음 리스트에서 60점 이상인 점수만 추출해라
scores = [45, 78, 92, 51, 88, 63, 39, 100]
new_list=[score for score in scores  if score>=60]
print(new_list)
```

```python
#문제2 enumerate()를 사용해서 1번부터 출력해라
machines = ["Robot-A", "Robot-B", "Robot-C", "Robot-D"]
for i, machine in enumerate(machines, start=1):
    print(i, machine)
```

```python
#문제3 두 리스트를 zip()으로 묶어서 출력해라
machine_ids = [101, 102, 103, 104]
temperatures = [72, 91, 105, 83]
for machine, temperature in zip(machine_ids, temperatures):
    print('Machine', machine,':', temperature)
```

```python
#문제4 함수 만들기 온도를 입력받아서 온도 위험 점수를 반환하는 함수를 만들어라
def calculate_temperature_score(temperature):
    # 작성
    if temperature<80:
        return 0
    elif temperature<=89:
        return 1
    elif temperature<=99:
        return 2
    else:
        return 3
# test
print(calculate_temperature_score(75))
print(calculate_temperature_score(85))
print(calculate_temperature_score(95))
print(calculate_temperature_score(105))
```

```python
#문제5 여러 센서의 위험 점수 계산
machine = {
    "id": 101,
    "temperature": 95,
    "pressure": 55,
    "vibration": 7.5
}
def calculate_temperature_score(temperature):
    if temperature<80:
        return 0
    elif temperature<=89:
        return 1
    elif temperature<=99:
        return 2
    else:
        return 3

def calculate_pressure_score(pressure):
    if pressure<40:
        return 0
    elif pressure<=49:
        return 1
    elif pressure<=59:
        return 2
    else:
        return 3

def calculate_vibration_score(vibration):
    if vibration<5:
        return 0
    elif vibration<=6.9:
        return 1
    elif vibration<=8.9:
        return 2
    else:
        return 3

temp=calculate_temperature_score(machine["temperature"])
pres=calculate_pressure_score(machine["pressure"])
vibr=calculate_vibration_score(machine["vibration"])
total_score=temp+pres+vibr
print("Temperature Score:",temp)
print("pressure Score:",pres)
print("vibration Score:",vibr)
print("Total Score:",total_score)
```

```python
#문제6 최종 상태 판단 함수
def determine_status(total_score):
    if total_score<=2:
        return "SAFE"
    elif total_score<=5:
        return "CHECK"
    else:
        return "DANGER"

dete=determine_status(total_score)

print("Temperature Score:",temp)
print("pressure Score:",pres)
print("vibration Score:",vibr)
print("Total Score:",total_score)
print("Status:", dete)
```
