```python
#1 list안에 dictionary
machines = [
    {"id": 1, "temperature": 70},
    {"id": 2, "temperature": 95},
    {"id": 3, "temperature": 105}
]
```

```python
#2 dictionart 값 수정
machine={
    "id":101,
    "temperature":95
}
machine["temperature"]=105
print(machine)
```

```python
#3 dictionary에 데이터 추가
machine["pressure"]=55
print(machine)
```

```python
#4 dictionary 데이터 삭제
del machine["pressure"]
print(machine)
```

```python
#5 데이터 중첩
factory = {
    "name": "Factory-A",
    "machines": [
        {"id": 101, "temperature": 80},
        {"id": 102, "temperature": 95},
        {"id": 103, "temperature": 105}
    ]
}
```

```python
print(factory["machines"])
print(factory["machines"][0])
print(factory["machines"][0]["temperature"])
```

```python
#문제1 Dictionary 수정
machine = {
    "id": 101,
    "temperature": 85,
    "pressure": 45,
    "status": "SAFE"
}
# 수정
machine["temperature"]=105
machine["pressure"]=60
machine["status"]="DANGER"
print(machine)
```

```python
#문제2 dictionary에 데이터 추가
machine = {
    "id": 101,
    "temperature": 85,
    "pressure": 45,
    "status": "SAFE"
}
#추가
machine["pressure"]=55
machine["vibration"]=7.2
machine["status"]="CHECK"
print(machine)
```
