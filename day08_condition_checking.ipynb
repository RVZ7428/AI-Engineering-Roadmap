```python
#any() 하나라도 참이면 True
temperatures = [72, 91, 105, 83, 110]

result = any(
    temperature >= 100
    for temperature in temperatures
)

print(result)
```

```python
#all() 전부 참이어야 True
temperatures = [72, 91, 105, 83, 110]

result = all(
    temperature < 100
    for temperature in temperatures
)

print(result)
```

```python
#Dictionary 데이터에 적용
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105}
]

has_danger = any(
    machine["temperature"] >= 100
    for machine in machines
)

print(has_danger)
```

```python
#문제1 위험 온도가 하나라도 있는가? 위험온도 : 100
temperatures = [72, 81, 95, 103, 88]
result=any(
    temperature>100
    for temperature in temperatures
)
print(result)
```

```python
#문제2 모든 기계가 안전한가? 모든 온도가 100 미만인지 검사
temperatures = [72, 81, 95, 83, 88]
result=all(
    temperature<100
    for temperature in temperatures
)
print(result)
```

```python
#문제3 모든 점수가 합격 기준인가? 합격기준 : 80
scores = [85, 92, 78, 96]
result=all(
    score>=80
    for score in scores
)
print(result)
```

```python
#문제4 기계 Dictionary 검사 온도가 100 이상인 기계가 있는지 검사
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105}
]
result=any(
    machine["temperature"]>=100
    for machine in machines
)
print(result)
```

```python
#문제5 함수 만들기 온도가 100 이상인 기계가 있으면 True 
def has_dangerous_machine(machines):
    # 작성
    result=any(
        machine["temperature"]>=100
        for machine in machines
        )
    return result
    
```
