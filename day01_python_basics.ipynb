```python
# 문제1 함수 완성하기
def calculate_average(scores):
    #작성
    s=0
    for score in scores:
        s+=score
    return s/len(scores)
    
#입력
scores=calculate_average([80, 90, 70, 100])
#결과 85.0
print(scores)
```

```python
# 문제3 공장에 다음과 같은 데이터가 들어온다
machines = [
    {"id": 1, "temperature": 70},
    {"id": 2, "temperature": 95},
    {"id": 3, "temperature": 105},
    {"id": 4, "temperature": 82},
    {"id": 5, "temperature": 110}
]
# 온도 기준을 다음과 같이 정한다
# 80 미만     → SAFE
# 80~99       → CHECK
# 100 이상    → DANGER

# 각 설비의 상태를 출력해라
for machine  in machines:
    temp = machine["temperature"]
    machine_id=machine["id"]

    if(temp<80):
        status="SAFE"
    elif(temp<100):
        status="CHECK"
    else:
        status="DANGER"

    print(f"Machine {machine_id}: {status}")
```

```python
# 문제4 함수+데이터 처리
def check_machine(machine):
    temperature = machine["temperature"]

    if temperature < 80:
        status = "SAFE"
    elif temperature < 100:
        status = "CHECK"
    else:
        status = "DANGER"

    return status


# 예시
# machine = {"id": 3, "temperature": 105} 
# print(check_machine(machine))
# 결과 DANGER
```

```python
def check_machine(machine):
    temp = machine["temperature"]

    if temp < 80:
        return "SAFE"
    elif temp < 100:
        return "CHECK"
    else:
        return "DANGER"


machines = [
    {"id": 1, "temperature": 70},
    {"id": 2, "temperature": 95},
    {"id": 3, "temperature": 105},
    {"id": 4, "temperature": 82},
    {"id": 5, "temperature": 110}
]

for machine in machines:
    status = check_machine(machine)
    print(f"Machine {machine['id']}: {status}")
```
