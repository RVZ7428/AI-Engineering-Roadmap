```python
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 95},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 110}
]
```

```python
def get_dangerous_machines(machines):
    result = []

    for machine in machines:
        if machine["temperature"] >= 100:
            result.append(machine)

    return result
```

```python
dangerous = get_dangerous_machines(machines)
print(dangerous)
```

```python
def get_machine_ids(machines):
    return [machine["id"] for machine in machines]
```

```python
dangerous = get_dangerous_machines(machines)
ids = get_machine_ids(dangerous)

print(ids)
```

```python
#문제1 위험 기계 찾기
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 95},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 110}
]

def get_dangerous_machines(machines):
    result=[]
    for machine in machines:
        if machine["temperature"]>=100:
            result.append(machine) # 딕셔너리 전체라 추출 암함
    return result
```

```python
#문제2 ID만 추출하기
def get_machine_ids(machines):
    result=[]
    for machine in machines:
        result.append(machine["id"])
    return result
```

```python
#문제3 함수 연결하기
dangerous=get_dangerous_machines(machines)
print(dangerous)
ids=get_machine_ids(dangerous)
print(ids)
```

```python
#문제4 위험 기계 수 계산하기
def count_dangerous_machines(machines):
    result=[]
    c=0
    for machine in machines:
        if machine["temperature"]>=100:
            result.append(machine)
    c=len(result)
    return c
```

```python
print(count_dangerous_machines(machines))
```
