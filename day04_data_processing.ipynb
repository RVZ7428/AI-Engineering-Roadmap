```python
#문제1 함수 연결
# 다음 함수를 완성해라
def add_ten(number):
    return number + 10

def multiply_two(number):
    return number * 2

# 그리고 두 함수를 연결해서 5 -> +10 -> *2 -> 결과 예상 : 30
result=multiply_two(add_ten(5))
print(result)
```

#문제2 점수 -> 등급
조건
90 이상 → A
80 이상 → B
70 이상 → C
60 이상 → D
60 미만 → F

```python
def get_grade(score):
    if score>=90:
        return "A"
    elif score>=80:
        return "B"
    elif score>=70:
        return "C"
    elif score>=60:
        return "D"
    else:
        return "F"
print(get_grade(95))
print(get_grade(82))
print(get_grade(73))
print(get_grade(65))
print(get_grade(40))
```

```python
#3
students = [
    {"name": "철수", "score": 95},
    {"name": "영희", "score": 82},
    {"name": "민수", "score": 73},
    {"name": "지수", "score": 58}
]

for student in students:
    grade = get_grade(student["score"])
    print(f"{student['name']}: {student['score']}점 {grade}")
```

```python
#문제4 함수로 데이터 필터링
# 온도가 100 이상인 기계의 ID만 리스트로 반환
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
def find_dangerous_machines(machines):
    dangerous_ids = []

    for machine in machines:
        if machine["temperature"] >= 100:
            dangerous_ids.append(machine["id"])

    return dangerous_ids
    

print(find_dangerous_machines(machines))
```

```python
#문제5 함수 조합
#온도 점수
def calculate_temperature_score(temperature):
    if temperature<80:
        return 0
    elif temperature<90:
        return 1
    elif temperature<100:
        return 2
    else:
        return 3

def determine_status(score):
    if score==0:
        return "SAFE"
    elif score==1:
        return "CHECK"
    else:
        return "DANGER"

machine = {
    "id": 101,
    "temperature": 95
}
print(determine_status(calculate_temperature_score(machine["temperature"])))
```
