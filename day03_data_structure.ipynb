```python
#문제5 중첩 데이터
factory = {
    "name": "Factory-A",
    "machines": [
        {"id": 101, "temperature": 80},
        {"id": 102, "temperature": 95},
        {"id": 103, "temperature": 110}
    ]
}
# 모든 기계의 id 출력
for machine in factory["machines"]:
    print(machine["id"])

# 모든 기계의 temperature 출력
for machine in factory["machines"]:
    print(machine["temperature"])

# 온도가 100 이상인 기계의 ID 출력
for machine  in factory["machines"]:
    if(machine["temperature"]>=100):
        print(machine["id"])
# 오류 string indices must be integers, not 'str'

```

```python
#문제6 함수로 만들기
def find_dangerous_machines(machines):
    new_list=[]
    #mahines를 받아서 온도가 100 이상인 기계의 id리스트로 반환해라
    for machine in machines:
        if(machine["temperature"]>=100):
            new_list.append(machine["id"])
    return new_list

machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 110}
]
dangerous = find_dangerous_machines(machines)

print(dangerous)
```
