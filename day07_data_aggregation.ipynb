```python
#5 실전 데이터에 적용하기
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
temperatures = [
    machine["temperature"] for machine in machines
]

average = sum(temperatures) / len(temperatures)

print(temperatures)
print(average)
```

```python
#문제3 Dictionary에서 온도 추출
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
list=[machine["temperature"] for machine in machines]
print(list)
```

```python
#문제5 최고 온도인 기계 찾기
machines = [
    {"id": 101, "temperature": 72},
    {"id": 102, "temperature": 91},
    {"id": 103, "temperature": 105},
    {"id": 104, "temperature": 83},
    {"id": 105, "temperature": 110}
]
max_temperature = max(
    machine["temperature"]
    for machine in machines
)

for machine in machines:
    if machine["temperature"] == max_temperature:
        print(machine["id"])

print(max_temperature)
```

```python
#문제6 함수로 만들기
def calculate_average_temperature(machines):
    temp=[machine["temperature"] for machine in machines]
    tempMean=sum(temp)/len(temp)
    return tempMean
```
